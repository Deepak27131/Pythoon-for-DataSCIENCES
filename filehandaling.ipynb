## File Handaling 
###### Bsaic function and methopd or:
- open()
- read()  
- close()

###### File object = open(file_name[,acces_mode][,buffering])

- Fileopen = open("filename","r")
- Fileopen = close()
- fileopen.read()
- fileopen.readline()
- fileopen.write()

```python
file = open("test.txt", 'r')
print(file)
```

```python
# line by line open the test file
for line in file:
    print(line)
```

```python
file = open("test.txt",'r')
print(file.read())
```

```python
## Safe & best method (recommended)

with open("test.txt",'r') as f:
    data = f.read()
    print(data)
```

```python
file = open("test.txt",'r')
print(file.readline())
```

```python
file = open("test.txt",'r')
print(file.read(20))
```

```python
file = open("test.txt",'r')
print(file.read(60))
```

```python
file = open("test.txt",'r')
print(file.readline(80))
```

### Writing file

```python
file = open("test.txt", 'w')
file.write("This is a write operation")
file.write("\n")
file.write("This time to learn the file handling")
file.write("\n")
file.write("Thanks you for joining the session")
file.close()
```

```python
print(file.read())
```

```python
with open("test.txt",'r') as f:
    data = f.read()
    print(data)
```

#### Append operation

```python
with open("test.txt",'a') as file:
    file.write("\n")
    file.write("This is append the write text")
```
