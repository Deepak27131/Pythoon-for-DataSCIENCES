```python
## itrate over a Dictionary and print each key-value pair

student = {1:["John",23],2:["Alice",22],3:["Bob",24],4:["Eve",21],5:["Charlie",23]}

for key,values in student.items():
    print("Key:",key)
    print("Values:",values)
```


### comprehensions

```python
list1 = ["Automobile","Finance","Health","Education","Retail"]

list2=[]

for i in list1:
    list2.append(i)
print(list2)

```

```python
list1 = ["Automobile","Finance","Health","Education","Retail"]

list2=[]

for i in list1:
    list2.append(len(i))
print(list2)
```

```python
list1 = ["Automobile","Finance","Health","Education","Retail"]

lst2 = [len(i) for i in list1]
print(lst2)
```

```python
list1 = ["Automobile","Finance","Health","Education","Retail"]

lst2 = {i:len(i) for i in list1}
print(lst2)
```
