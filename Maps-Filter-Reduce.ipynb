```python
# temps = [{"Mumbai",35},{"Berlin",12},{"Tokyo",25},{"Sydney",30},{"Hong Kong",31}]

temps = [("Delhi", 30), ("Mumbai", 32), ("Chennai", 35)]
type(temps)
```

```python
map(cel_to_f,temps)
```

### Reduce Function

reduc(f,data)

```python
## multiply all item in a loistr

from functools import reduce
data = [1,2,3,4,5]
```

```python
multiplier = lambda x,y:x*y

```

```python
reduce(multiplier,data)
```

```python
data = [1,2,3,4,5]
prev = data[0]

for i in data:
    i = i*prev
    prev = i
```
