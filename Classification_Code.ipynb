```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges,errors='coerce')
```

```python
df.info()
```

```python
df.dropna(how='any',inplace=True)
```

```python
df.head(5)
```

```python
x = df.drop(['customerID','Churn'],axis=1)
```

```python
y = df.Churn.values
```

```python
x
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
y = encoder.fit_transform(y)

print(y)  # Output: [1 0 1 0]
```

```python
y
```

```python
x.columns
```

```python
x = pd.get_dummies(x,columns=['gender', 'Partner', 'Dependents', 'tenure',
       'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
       'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod',
       ] ,drop_first=True,dtype='int')
```

```python
x
```

```python
# splititng data into train test split

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y ,test_size=0.25,random_state=42)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
# Feature Scailing
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
```

```python
x_train_sc = sc.fit_transform(x_train)
x_test_sc = sc.transform(x_test)
```

```python
x_train_sc
```

```python
x.head()
```

## KNN Classifier

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)

model.fit(x_train_sc,y_train)


y_pred = model.predict(x_test_sc)


# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred ,y_test)*100
```

```python
confusion_matrix(y_pred,y_test)
```

```python
from sklearn.metrics import recall_score,precision_recall_curve,classification_report
```

```python
print(classification_report(y_pred,y_test))
```

```python
from sklearn.metrics import roc_auc_score

print(roc_auc_score(y_pred,y_test))
```

```python

print(recall_score(y_pred,y_test))

print(precision_recall_curve(y_pred,y_test))
```

# Naive Bayes Classifier

```python
from sklearn.naive_bayes import  GaussianNB

GD = GaussianNB()

GD.fit(x_train_sc,y_train)


y_pred = GD.predict(x_test_sc)


# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred ,y_test)*100
```

# SVM

```python
from sklearn.svm import SVC

model_svc =SVC()

model_svc.fit(x_train_sc,y_train)


y_pred = model_svc.predict(x_test_sc)


# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred ,y_test)*100
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

model_lr =LogisticRegression( )

model_lr.fit(x_train_sc,y_train)


y_pred_lr = model_lr.predict(x_test_sc)


# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred_lr ,y_test)*100
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

model_rf =RandomForestClassifier(n_estimators=10,max_depth=5 )

model_rf.fit(x_train_sc,y_train)


y_pred_rf = model_rf.predict(x_test_sc)


# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred_rf ,y_test)*100
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

model_dc =DecisionTreeClassifier(criterion='entropy',splitter='random',max_features=10,max_depth=5)

model_dc.fit(x_train_sc,y_train)


y_pred_dc = model_dc.predict(x_test_sc)


# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred_dc ,y_test)*100
```
