# User-Define Functions
Unique block of code thant is only executed when call to perform a specific task:

#### Even or odd

```python
def even_odd(num):
    if num% 2==0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
```

```python
even_odd(10)
```

```python
even_odd(11)
```

### Factorial

```python
def factorial(n):
    if n ==0 or n==1:
        print("Factorial is 1")
    else:
        fact =1
        for i in range(2,n+1):
            fact = fact * i
        print(f"Factorial of {n} is {fact}")
```

```python
factorial(5)
```

```python
## other methods to find factorial
def fact_number(n):
    fact = 1
    for i in range(1,n+1):
        fact = fact * i
    return fact
```

```python
fact_number(5)
```

##### Sum of n natural numbers

```python
def sum_natural(n):
    sum = 0
    for i in range(1,n+1):
        sum += i
    return sum
```

```python
sum_natural(5)
```

## Lambda Function

- Small anonymous function to make a develop-er lifer easier
- Can take any number of arguments, but can only have one expression

lambda arguments:expression

```python
lambda arguments: expression
```

```python
x = lambda a:a+10
```

```python
print(x(5 ))
```

```python
print(x(25))
```

```python
x = lambda a:"even" if a%2==0 else "odd"
print(x(10))
```

```python
x = lambda a,b:a+b
print(x(5,10))
```

```python
x = lambda a,b:a*b
print(x(5,10))
```
