# Pandas 

- Pandas is a Python library for data analysis that provides powerful tools to work with structured data. At its core, it revolves around two main data structures:

### 1. Series
- A one-dimensional labeled array (like a single column in Excel).

- Can hold any data type: integers, floats, strings, or even Python objects.

- Each value has an index label, which makes data retrieval fast and intuitive.

### 2. DataFrame
- A two-dimensional table (like a spreadsheet or SQL table).

- Consists of multiple Series aligned by a common index.

- Each column can have a different data type (numeric, text, categorical).

- Rows are indexed, which allows easy selection, filtering, and joining.

- To load the pandas package and start working with it, import the package. The community agreed alias for pandas is pd, so loading pandas as pd is assumed standard practice for all of the pandas documentation.
