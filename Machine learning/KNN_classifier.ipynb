```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("D:\Pythoon for DataSCIENCES\Machine learning\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges,errors='coerce')
```

```python
df.info()
```

```python
df.dropna(how='any',inplace=True)
```

```python
df.head(5)
```

```python
x = df.drop(['customerID','Churn'],axis=1)
```

```python
y = df.Churn.values
```

```python
x
```

```python
y
```

```python
x.columns
```

```python
x = pd.get_dummies(x,columns=['gender', 'Partner', 'Dependents', 'tenure',
       'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
       'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod',
       ] ,drop_first=True,dtype='int')
```

```python
x
```

```python
# splititng data into train test split

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y ,test_size=0.25,random_state=42)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
# Feature Scailing
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
```

```python
x_train_sc = sc.fit_transform(x_train)
x_test_sc = sc.transform(x_test)
```

```python
x_train_sc
```

```python
x.head()
```

## KNN Classifier

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)

model.fit(x_train_sc,y_train)
```

```python
y_pred = model.predict(x_test_sc)
```

```python
y_pred
```

```python
y_test
```

```python
# check accuracy
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix

accuracy_score(y_pred ,y_test)*100
```

```python
confusion_matrix(y_pred,y_test)
```

```python
### check new data 

x_test.head(1)
```

# Disicion tree

```python
from sklearn.tree import DecisionTreeClassifier

dc = DecisionTreeClassifier()
dc.fit(x_train_sc,y_train)
```

```python
y_pred_dc = dc.predict(x_test_sc)
print(accuracy_score(y_pred_dc,y_test)*100)
print(confusion_matrix(y_pred_dc,y_test))
```

```python
## visulizating the decision tree

```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))
plot_tree(dc,filled=True,feature_names=x_train.columns,class_names=True)
plt.show()
```

```python
# using some hyperparameter 

from sklearn.tree import DecisionTreeClassifier

dc = DecisionTreeClassifier(max_depth=3)
dc.fit(x_train_sc,y_train)

y_pred_dc = dc.predict(x_test_sc)
print(accuracy_score(y_pred_dc,y_test)*100)
print(confusion_matrix(y_pred_dc,y_test))
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))
plot_tree(dc,filled=True,feature_names=x_train.columns,class_names=True)
plt.show()
```

```python
from sklearn.tree import export_graphviz
import graphviz

export_graphviz(dc ,out_file='tree.dot',feature_names=x_train.columns,class_names=True,filled=True,proportion=True,special_characters=True)


with open('tree.dot') as f:
    dot_graph = f.read()
graph = graphviz.Source(dot_graph,format='png')
graph
```
