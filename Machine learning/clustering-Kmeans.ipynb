```python
# Importing library 
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('mall.csv')
df.head()
```

```python
x = df.iloc[:,[3,4]].values
x
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
    
plt.plot(range(1,11),wcss)
plt.title("Elbow Method")
plt.xlabel('Clusters')
plt.ylabel('WCSS')
```

```python

kmeans = KMeans(n_clusters=5)
y_kmeans = kmeans.fit_predict(x)
```

```python
# Visulizating the cluster 
plt.Figure(figsize=(10,15))
plt.scatter(x[y_kmeans == 0, 0], x[y_kmeans == 0, 1], s=100, c='red', label='Cluster 1', edgecolor='k')
plt.scatter(x[y_kmeans == 1,0],x[y_kmeans == 1,1], s = 100, c ='blue', label = ' cluster_2')
plt.scatter(x[y_kmeans == 2,0],x[y_kmeans == 2,1], s = 100, c ='green', label = ' cluster_3')
plt.scatter(x[y_kmeans == 3,0],x[y_kmeans == 3,1], s = 100, c ='yellow', label = ' cluster_4')
plt.scatter(x[y_kmeans == 4,0],x[y_kmeans == 4,1], s = 100, c ='hotpink', label = ' cluster_5')
plt.scatter(x[y_kmeans == 5,0],x[y_kmeans == 5,1], s = 100, c ='brown', label = ' cluster_6')
plt.scatter(x[y_kmeans == 6,0],x[y_kmeans == 6,1], s = 100, c ='cyan', label = ' cluster_5')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],s = 300,c='black',label = 'Centroid')

plt.title("Cluster of  client")
plt.xlabel('Annual Icome')
plt.ylabel('Spending Score (0-100)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[2,4],[4,2],[4,4],[2,2],[8,7],[8,9]])
kmeans = KMeans(n_clusters=2, random_state=0).fit(X)

print("Centroids:", kmeans.cluster_centers_)
print("Labels:", kmeans.labels_)

plt.scatter(X[:,0], X[:,1], c=kmeans.labels_, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], c='red', marker='x', s=200)
plt.title("K-Means Clustering")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Step 1: Define the dataset
X = np.array([[2,4], [4,2], [4,4], [2,2], [8,7], [8,9]])

# Step 2: Apply K-Means (K=2)
kmeans = KMeans(n_clusters=2, init='random', n_init=1, random_state=0)
labels = kmeans.fit_predict(X)
centroids = kmeans.cluster_centers_

# Step 3: Plot the data points and centroids
plt.figure(figsize=(8,6))

# Plot data points with cluster colors
scatter = plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=100, edgecolor='k', linewidth=2)

# Plot centroids
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='x', s=300, linewidth=5, label='Centroids')

# Annotate points
for i, (x, y) in enumerate(X):
    plt.text(x, y, f'P{i+1}', fontsize=12, ha='center', va='bottom')

# Add labels and title
plt.title('K-Means Clustering (K=2) - 6 Data Points', fontsize=14)
plt.xlabel('Feature 1 (X)')
plt.ylabel('Feature 2 (Y)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Print results
print("=== K-Means Results ===")
print("Data Points:\n", X)
print("Cluster Labels (0 or 1):\n", labels)
print("Centroids:\n", centroids)
```
