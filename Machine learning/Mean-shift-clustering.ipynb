```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import MeanShift
```

```python
df= pd.read_csv('mall.csv')
df.head(5)
```

```python
x = df.iloc[:,[3,4]].values
```

```python
# Fitting the mean shift clustering to the mall dataset

ms = MeanShift(bandwidth=20)

# Adjust the bandwith parameter to influences the number of cluster

ms.fit(x)
labels = ms.labels_
cluster_center = ms.cluster_centers_
```

```python
labels
```

```python
# visulization

n_clusters = len(np.unique(labels))

colors = ['red','blue','green','yellow','cyan','purple','pink']

for i in range(n_clusters):
    cluster = x[labels == i]
    plt.scatter(cluster[:,0],cluster[:,1] , s=100,c = colors[i], label = f'Cluster{i + 1}')
    
plt.title("Cluster ")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score ( 0 - 100)")
plt.grid()
plt.legend()
plt.show()
```
