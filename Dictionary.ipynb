# Defining a dictionary in Python
- Dictionary store key values pair elements
- The key is like an index ,its alwayscunique & immutable
- the values are object that coontain information
- Each key is follwed by a values separated by a colon
- tyhe value can be immutable, mutable and duplicated
- each key value pair is separated by a colon and enclosed by curly bracket

```python
my_dict = {'key1': 'value1', 'key2': 'value2', 'key3': 42}
print(my_dict)
```

```python
input_dict = {
    'name': 'Alice',
    'Profession': 'Software Engineer'}

if "label" in input_dict.keys():
    print(input_dict["label"])
else:
    input_dict["label"] = "NA"
    print(input_dict)
```
